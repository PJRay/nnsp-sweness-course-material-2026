![sword model](sword_model.png)

The simulation can be performed by selecting the **Sword_ODIN** instrument in the McStas application on the virtual desktop. There are 9 different perameters that control the simulation.

**chopper_mode** selects which wavelengths are available and the wavelength resolution we have in the instrument. The statistics of each chopper are as follows:

```text
chopper_mode=0: wavelength from 1.0 Å to 9.0 Å and a resolution of +/- 10%
chopper_mode=1: wavelength from 1.0 Å to 5.5 Å and a resolution of +/- 5%
chopper_mode=2: wavelength from 2.5 Å to 6.5 Å and a resolution of +/- 4%
chopper_mode=3: wavelength from 3.8 Å to 8.8 Å and a resolution of +/- 3%
chopper_mode=4: wavelength from 5.0 Å to 9.0 Å and a resolution of +/- 2%
chopper_mode=5: white beam with wavelength from ~0 Å to ~10 Å
```

**Lambda** allows us to select a specific wavelength in the allowed wavelength range we want to simulate.

**Sample** controls if we are simulating the background or the sample+background. Sample=1 means that the sample is in the beam. Sample=0 means that there is no sample in the beam.

**pinhole_diameter** controls the diameter of the pinhole, the allowed range are from 0.01 m to 0.1 m.

**pinhole_detector_distance** sets the distance from the pinhole to the detector in meters. The minimum distance is 10 m and the maximum is 25 m.

**pinhole_sample_distance** sets the distance from the pinhole to the center of the sample. The minimum distance is 1 m and the maximum is 25 m. Remember that the detector and the sample may not overlap, and that the detector should be behind the sample at all times.

**X_sample_pos** and **Y_sample_pos** translate the sword either horizontaly (x) or verticaly (y). The center of the sword is placed in x=0, y=0.

**angle** rotates the sample around the y axis.

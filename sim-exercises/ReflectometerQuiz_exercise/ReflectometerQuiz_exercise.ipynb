The figure below is an illustration of the reflectometer instrument's layout. Each of the components illustrated here are parts of the simulation and, as such, they are regarded as instrument parameters that will influence the simulation results when adjusted.

![instrument layout](instrument_layout.png)

The instrument parameters are following the format **parameter_name: [units] Explanation**. In order to perform the simulations, select the **reflectometer_alignment** instrument file in the McStas application on the virtual desktop. Under the list of parameters you can see a screenshot of the McStas simulation interface of the reflectometer, which you will be using. Note the predefined default values.

Please take the time to study and understand what they do. Make sure to ask if you have any questions.

```text
lambda_min: [AA]   Minimum wavelength from source
lambda_max: [AA]   Maximum wavelength from source
sampletranslation: [m]    Sample translation orthogonal to incoming beam
slitwidth: [m]    Width of slit pinholes
slitheight: [m]    Height of slit pinholes
dist_source2slit: [m]    Distance between source and first slit
dist_slit2slit: [m]    Distance between slits
dist_slit2sample: [m]    Distance between second slit and sample
dist_sample2detector: [m]    Distance between sample and detector
sampletype: [1]    Sample type: 0 none, 1 mirror, 2+ multilayer
MR_Qc: [AA^-1]   Critical Q-vector length of mirror sample
sampleangle: [deg]  Rotation angle of sample aka theta
detectorangle: [deg]  Rotation angle of detector window aka 2 theta
beamstop: [1]    If 0 the beamstop is out, if 1 it is in
```
